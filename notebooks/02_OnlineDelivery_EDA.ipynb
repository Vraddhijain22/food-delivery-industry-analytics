```python
#STEP 1: Import Libraries
import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

sns.set_style("whitegrid")
```

```python
#STEP 2: Load Dataset
delivery = pd.read_csv("../data/onlinedeliverydata.csv")
```

```python
delivery.shape
```

```python
#STEP 3: Preview Dataset
delivery.head()
```

```python
#TEP 4: Dataset Information
delivery.info()
```

```python
delivery.columns.tolist()
```

```python
#STEP 5: Check for Missing Values
delivery.isnull().sum()
```

```python
#STEP 6: Check for Duplicate Values
delivery.duplicated().sum()
```

```python
#STEP 7: Drop Duplicate Values if any
delivery = delivery.drop_duplicates()

print(delivery.shape)
```

```python
delivery.columns.tolist()
```

```python
# STEP 8: Age Distribution
plt.figure(figsize=(8,5))

sns.histplot(
    delivery['Age'],
    bins=10,
    kde=True,
    color='orange'
)

plt.title('Customer Age Distribution')
plt.show()
```

Most food delivery users belong to younger age groups.

```python
#STEP 9: Gender Distribution
plt.figure(figsize=(6,4))

sns.countplot(
    x='Gender',
    data=delivery
)

plt.title('Gender Distribution')

plt.show()
```

```python
#STEP 10: Occupation Analysis
plt.figure(figsize=(10,6))

delivery['Occupation'] \
.value_counts() \
.plot(kind='bar')

plt.title('Occupation Distribution')

plt.show()
```

Students generally form the largest customer segment.

```python
#STEP 11: Monthly Income Analysis

plt.figure(figsize=(12,6))

delivery['Monthly Income'] \
.value_counts() \
.plot(kind='bar')

plt.title('Monthly Income Distribution')

plt.show()
```

```python
#STEP 12: Preferred Ordering Method

# Medium (P1)
plt.figure(figsize=(10,5))

delivery['Medium (P1)'] \
.value_counts() \
.plot(kind='bar')

plt.title('Preferred Ordering Method')

plt.show()
```

Food delivery apps dominate compared to direct ordering.

```python
#STEP 13: Meal Preference
plt.figure(figsize=(8,5))

delivery['Meal(P1)'] \
.value_counts() \
.plot(kind='bar')

plt.title('Preferred Meal')

plt.show()
```

```python
#STEP 13: Food Preference
plt.figure(figsize=(12,6))

delivery['Perference(P1)'] \
.value_counts() \
.plot(kind='bar')

plt.title('Food Preferences')

plt.show()
```

```python
#STEP 15: Customer Reasons for Using Delivery Apps
factors = [
    'Ease and convenient',
    'Time saving',
    'More restaurant choices',
    'Easy Payment option',
    'More Offers and Discount',
    'Good Food quality'
]

for col in factors:
    print("\n")
    print(col)
    print(delivery)
```

```python
#STEP 16: Visualize Key Factors
factor_counts = {}

for col in factors:
    factor_counts[col] = (
        delivery[col]
        .value_counts()
        .idxmax()
    )

factor_counts
```

```python
# STEP 17: Influence of Ratings
plt.figure(figsize=(8,5))

delivery['Influence of rating'] \
.value_counts() \
.plot(kind='bar')

plt.title('Influence of Ratings')

plt.show()
```

Ratings strongly influence restaurant selection decisions.

```python
#STEP 18: Review Analysis

delivery['Reviews'].value_counts()
```

```python
plt.figure(figsize=(8,5))

delivery['Reviews'] \
.value_counts() \
.plot(kind='bar')

plt.title('Customer Reviews')

plt.show()
```

```python
#STEP 19: Correlation Heatmap
numeric_cols = delivery.select_dtypes(
    include=['int64','float64']
)

```

```python
plt.figure(figsize=(10,8))

sns.heatmap(
    numeric_cols.corr(),
    annot=True,
    cmap='coolwarm'
)

plt.title('Correlation Matrix')

plt.show()
```
