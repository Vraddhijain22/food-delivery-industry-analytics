```python
# Step 1: Import Libraries

import pandas as pd
import numpy as np

import matplotlib.pyplot as plt
import seaborn as sns

print("Libraries Loaded Successfully")
```

```python
# Step 2: Load the Clean Dataset

zomato = pd.read_csv("../data/cleaned_zomato.csv")

delivery = pd.read_csv("../data/cleaned_delivery.csv")

print("Zomato Shape:", zomato.shape)
print("Delivery Shape:", delivery.shape)
```

Executive Summary Section

```python
# Step 3: Restaurant KPIs
print("========== RESTAURANT KPIs ==========")

print("Total Restaurants:",
      zomato['name'].nunique())

print("Average Rating:",
      round(zomato['rate'].mean(),2))

print("Average Cost For Two:",
      round(
          zomato['approx_cost(for two people)'].mean(),
          2
      ))

print("Total Votes:",
      zomato['votes'].sum())
```

```python
#Step 4: Customer KPIs
print("========== CUSTOMER KPIs ==========")

print("Total Customers:",
      len(delivery))

print("Average Age:",
      round(delivery['Age'].mean(),2))

print("Most Common Occupation:",
      delivery['Occupation'].mode()[0])

print("Most Common Ordering Method:",
      delivery['Medium (P1)'].mode()[0])
```

Restaurant Analytics Summary

```python
#Step 5: Most Popular Location and Cuisine
print("Most Popular Location:")

print(
    zomato['location']
    .value_counts()
    .idxmax()
)

print("\nMost Popular Cuisine:")

print(
    zomato['cuisines']
    .value_counts()
    .idxmax()
)
```

```python
#Step 6: Most Popular Restaurant
top_restaurant = (
    zomato.groupby('name')['votes']
    .sum()
    .idxmax()
)

print("Most Popular Restaurant:")

print(top_restaurant)
```

Customer Analytics Summary

```python
#Step 7: Top Occupations

print("Top Occupation:")

print(
    delivery['Occupation']
    .value_counts()
    .idxmax()
)
```

```python
#Step 8: Preferred Meal
print("Most Preferred Meal:")

print(
    delivery['Meal(P1)']
    .value_counts()
    .idxmax()
)
```

```python
#Step 9: Food Category Preferences
print("Preferred Food Category:")

print(
    delivery['Perference(P1)']
    .value_counts()
    .idxmax()
)
```

Business Drivers

```python
#Step 10: Business

business_factors = [
    'Ease and convenient',
    'Time saving',
    'More restaurant choices',
    'Easy Payment option',
    'More Offers and Discount',
    'Good Food quality'
]

for factor in business_factors:
    
    print("\n"+factor)
    
    print(
        delivery[factor]
        .value_counts()
        .head(1)
    )
```

Customer Decision Factors

```python
#Step 11: Influence of Rating
print("Influence of Rating:")

print(
    delivery['Influence of rating']
    .value_counts()
)
```
